### Check Port Connectivity (Simulated)

You can try to use `nc` (netcat) or `telnet` (if available, sometimes `nc` is more common) to check if a specific port on a server is open. However, `telnet` is often not installed by default in Colab, and `nc` can be used. This simulates trying to open a connection to the database port.

```python
get_ipython().system('apt-get update && apt-get install -y netcat')
get_ipython().system('nc -zv db.ckezmybqqfngouvzvaym.supabase.co 5432')
```

```python
import re

# Extract the hostname from the DB_URL
match = re.search(r'@([^:]+):', DB_URL)
if match:
    supabase_host = match.group(1)
    print(f"Pinging Supabase host: {supabase_host}")
    get_ipython().system(f'ping -c 4 {supabase_host}')
else:
    print("Could not extract Supabase host from DB_URL.")
```

The output of `nc` will indicate if the connection to the host and port was successful or if it was refused/timed out. If it says `Connection refused` or `Connection timed out`, it could mean the Supabase server is not accessible from Colab on that specific port, possibly due to a firewall on the server side, or network routing issues.
