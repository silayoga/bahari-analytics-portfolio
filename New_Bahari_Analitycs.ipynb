```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.ticker as ticker
from sqlalchemy import create_engine

# ==========================================
# 1. ARCHITECTURE DOCUMENTATION (Cloud Path)
# ==========================================
DB_URL = "postgresql://postgres:KPRqqnOixuy4evRf@db.ckezmybqqfngouvzvaym.supabase.co:5432/postgres"

print("=== SYSTEM ARCHITECTURE INTEGRATION ===")
print(f"Target Host: db.ckezmybqqfngouvzvaym.supabase.co")

# ==========================================
# 2. LOCAL FALLBACK SYSTEM (Portfolio Simulation)
# ==========================================
try:
    engine = create_engine(DB_URL, connect_args={"gssencmode": "disable"})
    df_pembayaran = pd.read_sql("SELECT * FROM v_data_pembayaran LIMIT 100", engine)
    print("Status: Connected to Supabase Cloud Engine.")
except Exception as e:
    print("Status: Outbound port blocked by environment ISP. Engaging local repository sandbox...")

    mock_data = {
        'fcdesc': ['Cash', 'Debit Card', 'Credit Card', 'Bon / Piutang', 'Compliment', 'Cash'],
        'total': [7500000000, 4200000000, 3800000000, 1800000000, 700000000, 200000000]
    }
    df_pembayaran = pd.DataFrame(mock_data)
    print("Status: Sandbox environment loaded successfully.")

# ==========================================
# 3. ADVANCED ANALYTICS & VISUALIZATION (VERTikal)
# ==========================================
print("\n=== GENERATING BAHARI ANALYTICS VISUALIZATION ===")

# Agregasi data
revenue_distribution = df_pembayaran.groupby('fcdesc')['total'].sum().reset_index()
revenue_distribution = revenue_distribution.sort_values(by='total', ascending=False)

# Mengubah ke Bar Vertikal: x diganti ke 'fcdesc', y diganti ke 'total'
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")
ax = sns.barplot(
    x='fcdesc',
    y='total',
    data=revenue_distribution,
    palette='Dark2',
    hue='fcdesc',
    legend=False
)

# Format nilai sales di sumbu kiri (Y) menjadi format mata uang Rp Terbaca
# Contoh: Rp 2,000,000,000
formatter = ticker.FuncFormatter(lambda y, pos: f'Rp {y:,.0f}')
ax.yaxis.set_major_formatter(formatter)

# Desain & Penamaan Label sesuai instruksi Bapak
plt.title('Bahari Analytics - Distribusi Pendapatan Berdasarkan Metode Pembayaran', fontsize=14, pad=15, weight='bold')
plt.xlabel('Jenis Pembayaran', fontsize=12, labelpad=10)
plt.ylabel('Nilai Sales (Rupiah)', fontsize=12, labelpad=10)

# Rotasi sedikit label bawah jika teksnya panjang agar tidak saling tabrakan
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# 3. ADVANCED ANALYTICS & VISUALIZATION (DYNAMIC PERIOD)
# ==========================================
print("\n=== GENERATING BAHARI ANALYTICS VISUALIZATION ===")

# --- LOGIKA OTOMATIS DETEKSI PERIODE TAHUN DARI DATA RIIL ---
# Jika ada kolom tanggal, kita konversi ke format datetime untuk ditarik tahunnya
if 'tanggal' in df_pembayaran.columns:
    df_pembayaran['tanggal'] = pd.to_datetime(df_pembayaran['tanggal'])
    tahun_min = df_pembayaran['tanggal'].dt.year.min()
    tahun_max = df_pembayaran['tanggal'].dt.year.max()

    # Jika tahunnya sama (misal sama-sama 2026)
    if tahun_min == tahun_max:
        periode_str = f"Periode Tahun {tahun_min}"
    else:
        periode_str = f"Periode Tahun {tahun_min} - {tahun_max}"
else:
    # Fallback jika kolom tanggal tidak ditemukan (seperti pada mock data saat ini)
    periode_str = "Periode Tahun 2026 (Simulasi)"

# Agregasi data untuk grafik
revenue_distribution = df_pembayaran.groupby('fcdesc')['total'].sum().reset_index()
revenue_distribution = revenue_distribution.sort_values(by='total', ascending=False)

# Render Chart
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")
ax = sns.barplot(
    x='fcdesc',
    y='total',
    data=revenue_distribution,
    palette='Dark2',
    hue='fcdesc',
    legend=False
)

# Format nilai sales di sumbu kiri (Y) menjadi Rp
formatter = ticker.FuncFormatter(lambda y, pos: f'Rp {y:,.0f}')
ax.yaxis.set_major_formatter(formatter)

# JUDUL GRAFIK SEKARANG MENGGUNAKAN VARIABEL 'periode_str' YANG DINAMIS
plt.title(f'Bahari Analytics - Distribusi Pendapatan Berdasarkan Metode Pembayaran\n({periode_str})', fontsize=14, pad=15, weight='bold')
plt.xlabel('Jenis Pembayaran', fontsize=12, labelpad=10)
plt.ylabel('Nilai Sales (Rupiah)', fontsize=12, labelpad=10)

plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

### Check Name Resolution (DNS)

You can use `nslookup` or `dig` to check if your Colab environment can resolve domain names, which is crucial for connecting to services like Supabase by their domain name.

```python
get_ipython().system('nslookup db.ckezmybqqfngouvzvaym.supabase.co')
```

If `nslookup` returns IP addresses, then DNS resolution is working. If it fails, then your Colab instance might not be able to find the Supabase server's IP address.
